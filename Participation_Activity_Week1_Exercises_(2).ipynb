## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
manual_mean = sum(salaries) / len(salaries)
print(f"Manual Mean: {manual_mean}")

import statistics
statistical_mean = statistics.mean(salaries)
print(f"Statistical Mean: {statistical_mean}")

assert manual_mean == statistical_mean, "Manual and statistical means do not match!"
print("Verification successful: Manual and statistical means match.")
```

### median

```python
sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

if n % 2 == 1:
    manual_median = sorted_salaries[n // 2]
else:
    mid1 = sorted_salaries[n // 2 - 1]
    mid2 = sorted_salaries[n // 2]
    manual_median = (mid1 + mid2) / 2

print(f"Manual Median: {manual_median}")

import statistics
statistical_median = statistics.median(salaries)
print(f"Statistical Median: {statistical_median}")

assert manual_median == statistical_median, "Manual and statistical medians do not match!"
print("Verification successful: Manual and statistical medians match.")
```

### mode

```python
from collections import Counter

salaries_counts = Counter(salaries)

if not salaries_counts:
    manual_mode = []
else:
    max_count = max(salaries_counts.values())
    manual_mode = [key for key, value in salaries_counts.items() if value == max_count]

print(f"Manual Mode: {manual_mode}")

import statistics
statistical_mode = statistics.multimode(salaries)
print(f"Statistical Mode: {statistical_mode}")

# Sort both lists for consistent comparison, as order doesn't matter for modes
assert sorted(manual_mode) == sorted(statistical_mode), "Manual and statistical modes do not match!"
print("Verification successful: Manual and statistical modes match.")
```

### sample variance
Remember to use Bessel's correction.

```python
mean_salaries = sum(salaries) / len(salaries)

# Calculate manual sample variance using Bessel's correction (n-1)
manual_sample_variance = sum([(x - mean_salaries) ** 2 for x in salaries]) / (len(salaries) - 1)
print(f"Manual Sample Variance: {manual_sample_variance}")

import statistics
statistical_sample_variance = statistics.variance(salaries)
print(f"Statistical Sample Variance: {statistical_sample_variance}")

assert abs(manual_sample_variance - statistical_sample_variance) < 1e-9, "Manual and statistical sample variances do not match!"
print("Verification successful: Manual and statistical sample variances match.")
```

### sample standard deviation
Remember to use Bessel's correction.

```python
import random
import math
import statistics

# Redefine salaries to ensure it's available in this cell
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]

mean_salaries = sum(salaries) / len(salaries)

# Calculate manual sample variance using Bessel's correction (n-1)
manual_sample_variance = sum([(x - mean_salaries) ** 2 for x in salaries]) / (len(salaries) - 1)

# Manual sample standard deviation is the square root of the sample variance
manual_sample_std_dev = math.sqrt(manual_sample_variance)
print(f"Manual Sample Standard Deviation: {manual_sample_std_dev}")

# Statistical sample standard deviation
statistical_sample_std_dev = statistics.stdev(salaries)
print(f"Statistical Sample Standard Deviation: {statistical_sample_std_dev}")

assert abs(manual_sample_std_dev - statistical_sample_std_dev) < 1e-9, "Manual and statistical sample standard deviations do not match!"
print("Verification successful: Manual and statistical sample standard deviations match.")
```

### interquartile range

```python
import math
import statistics

# Ensure salaries is defined for independent execution of this cell
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]

sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

# Manual calculation of Q1 (25th percentile) using linear interpolation for 'exclusive' method
p_q1 = 0.25
j_0based_q1 = p_q1 * (n + 1) - 1 # Calculate 0-based fractional index
k_q1 = int(j_0based_q1)        # Integer part of the index
f_q1 = j_0based_q1 - k_q1       # Fractional part of the index

# Perform linear interpolation between the two closest data points
manual_q1 = sorted_salaries[k_q1] + f_q1 * (sorted_salaries[k_q1+1] - sorted_salaries[k_q1])

# Manual calculation of Q3 (75th percentile) using linear interpolation for 'exclusive' method
p_q3 = 0.75
j_0based_q3 = p_q3 * (n + 1) - 1 # Calculate 0-based fractional index
k_q3 = int(j_0based_q3)        # Integer part of the index
f_q3 = j_0based_q3 - k_q3       # Fractional part of the index

# Perform linear interpolation between the two closest data points
manual_q3 = sorted_salaries[k_q3] + f_q3 * (sorted_salaries[k_q3+1] - sorted_salaries[k_q3])

manual_iqr = manual_q3 - manual_q1
print(f"Manual Interquartile Range (IQR): {manual_iqr}")

# Verification using the statistics module
# The default method for quantiles in statistics.quantiles is 'exclusive'
# which aligns with the method used above for indices.
quantiles = statistics.quantiles(salaries, n=4, method='exclusive')
statistical_q1 = quantiles[0]
statistical_q3 = quantiles[2]

statistical_iqr = statistical_q3 - statistical_q1
print(f"Statistical Interquartile Range (IQR): {statistical_iqr}")

assert abs(manual_iqr - statistical_iqr) < 1e-9, "Manual and statistical IQRs do not match!"
print("Verification successful: Manual and statistical IQRs match.")
```

### quartile coefficent of dispersion

```python
import statistics

# Using the manual_q1 and manual_q3 from the previous cell
manual_qcd = (manual_q3 - manual_q1) / (manual_q3 + manual_q1)
print(f"Manual Quartile Coefficient of Dispersion (QCD): {manual_qcd:.4f}")

# Using statistical_q1 and statistical_q3 for verification
statistical_qcd = (statistical_q3 - statistical_q1) / (statistical_q3 + statistical_q1)
print(f"Statistical Quartile Coefficient of Dispersion (QCD): {statistical_qcd:.4f}")

assert abs(manual_qcd - statistical_qcd) < 1e-9, "Manual and statistical QCD do not match!"
print("Verification successful: Manual and statistical QCD match.")
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import random
import statistics

# Generate a new list for demonstration, e.g., years of experience
random.seed(1) # Using a different seed for a different distribution
years_of_experience = [round(random.uniform(1, 30), 0) for _ in range(len(salaries))]

print("First 5 salaries:", salaries[:5])
print("First 5 years of experience:", years_of_experience[:5])

# Calculate means for both lists
mean_salaries = sum(salaries) / len(salaries)
mean_years_of_experience = sum(years_of_experience) / len(years_of_experience)

# Manual calculation of sample covariance
# Covariance = sum((xi - mean_x) * (yi - mean_y)) / (n - 1)
n_cov = len(salaries)
manual_covariance = sum([
    (salaries[i] - mean_salaries) * (years_of_experience[i] - mean_years_of_experience)
    for i in range(n_cov)
]) / (n_cov - 1)

print(f"\nManual Sample Covariance: {manual_covariance:.2f}")

# Verification using the statistics module
statistical_covariance = statistics.covariance(salaries, years_of_experience)
print(f"Statistical Sample Covariance: {statistical_covariance:.2f}")

assert abs(manual_covariance - statistical_covariance) < 1e-9, "Manual and statistical covariances do not match!"
print("Verification successful: Manual and statistical covariances match.")
```

### Pearson correlation coefficient ($\rho$)

```python
import math
import statistics

# Ensure salaries, years_of_experience, mean_salaries, mean_years_of_experience,
# manual_sample_std_dev, and manual_covariance are available from previous cells.

# Calculate manual sample standard deviation for years_of_experience
# First, calculate the sample variance for years_of_experience
manual_sample_variance_yoe = sum([(y - mean_years_of_experience) ** 2 for y in years_of_experience]) / (len(years_of_experience) - 1)
manual_sample_std_dev_yoe = math.sqrt(manual_sample_variance_yoe)

# Manual calculation of Pearson correlation coefficient
# r = Cov(X, Y) / (StdDev(X) * StdDev(Y))
manual_pearson_correlation = manual_covariance / (manual_sample_std_dev * manual_sample_std_dev_yoe)
print(f"Manual Pearson Correlation Coefficient: {manual_pearson_correlation:.4f}")

# Verification using the statistics module
statistical_pearson_correlation = statistics.correlation(salaries, years_of_experience)
print(f"Statistical Pearson Correlation Coefficient: {statistical_pearson_correlation:.4f}")

assert abs(manual_pearson_correlation - statistical_pearson_correlation) < 1e-9, "Manual and statistical Pearson correlations do not match!"
print("Verification successful: Manual and statistical Pearson Correlation Coefficients match.")
```
