## Part 1. Brief Review of Lambda Functions

A lambda expression creates a small anonymous function.

The standard Python syntax is:

```python
lambda parameters: expression
```

For example:

```python
square = lambda x: x ** 2
square(5)
```

returns `25`.

According to the Python documentation, a lambda expression creates a function object. A lambda can contain an expression, but it cannot contain regular statements such as a multi-line `for` loop or `if` statement.

PEP 312 proposed a shorter form of implicit lambda syntax in some situations. However, PEP 312 is currently marked **Deferred**, so this notebook uses the normal Python `lambda` syntax.

References:

- Python Lambda Expressions: https://docs.python.org/3/reference/expressions.html#lambda
- PEP 312 – Simple Implicit Lambda: https://peps.python.org/pep-0312/

## Part 3. Load the Student Performance Data Set



````python
import pandas as pd

DATA_PATH = REPO_DIR / "data" / "Student Performance Data.csv"

print("Data file exists:", DATA_PATH.exists())

students = pd.read_csv(DATA_PATH)

print("DataFrame shape:", students.shape)
students.head()
````

## Part 4. Inspect the Data



````python
print("Columns:")
print(students.columns.tolist())

print("\nData types:")
print(students.dtypes)
````

The next cell checks for missing values.

````python
students.isna().sum()
````

## Part 5. Create a Lambda Function That Returns Another DataFrame



````python
create_score_dataframe = lambda data: data.assign(
    **{
        "average score": lambda x: x[
            ["math score", "reading score", "writing score"]
        ].mean(axis=1)
    }
)
````

## Part 6. Apply the Lambda Function


````python
student_scores = create_score_dataframe(students)

print("Original DataFrame shape:", students.shape)
print("New DataFrame shape:", student_scores.shape)

student_scores.head()
````

### Interpretation

The lambda function successfully takes the original Student Performance DataFrame as input and returns another DataFrame.

The returned DataFrame contains all original student information plus a new `average score` column. This column summarizes each student's performance across the math, reading, and writing tests.

For example, if a student has scores of 72 in math, 72 in reading, and 74 in writing, the average score is approximately 72.67.

## Part 7. Verify That a New DataFrame Was Returned


````python
print("Original columns:", len(students.columns))
print("New columns:", len(student_scores.columns))

print("\nIs 'average score' in original DataFrame?")
print("average score" in students.columns)

print("\nIs 'average score' in new DataFrame?")
print("average score" in student_scores.columns)
````

## Part 8. View Summary Statistics for the New Variable



````python
student_scores["average score"].describe()
````

## Part 9. Optional: View Students with the Highest Average Scores



````python
student_scores.sort_values(
    by="average score",
    ascending=False
).head(10)
````
