# MSSP607 Week 5 Participation Activity


## Part 1. Clone the GitHub Repository


```python
from pathlib import Path
import os
import subprocess

REPO_URL = "https://github.com/kaixing189/MSSP607.git"
REPO_DIR = Path("/content/MSSP607")

if REPO_DIR.exists():
    subprocess.run(
        ["git", "-C", str(REPO_DIR), "pull"],
        check=True
    )
    print("Repository already exists. Latest changes were pulled.")
else:
    subprocess.run(
        ["git", "clone", REPO_URL, str(REPO_DIR)],
        check=True
    )
    print("Repository cloned successfully.")

os.chdir(REPO_DIR)
print("Current working directory:", os.getcwd())
```

## Part 2. Define the Data Folder and Check the Files



```python
DATA_DIR = REPO_DIR / "data"

required_files = [
    "aapl.csv",
    "amzn.csv",
    "fb.csv",
    "goog.csv",
    "nflx.csv"
]

for filename in required_files:
    file_path = DATA_DIR / filename
    print(f"{filename}: {file_path.exists()}")

assert all((DATA_DIR / filename).exists() for filename in required_files),     "One or more required CSV files are missing."
```

## Part 3. Import pandas and Read Each CSV File



```python
import pandas as pd

aapl = pd.read_csv(DATA_DIR / "aapl.csv")
amzn = pd.read_csv(DATA_DIR / "amzn.csv")
fb = pd.read_csv(DATA_DIR / "fb.csv")
goog = pd.read_csv(DATA_DIR / "goog.csv")
nflx = pd.read_csv(DATA_DIR / "nflx.csv")

print("AAPL shape:", aapl.shape)
print("AMZN shape:", amzn.shape)
print("FB shape:", fb.shape)
print("GOOG shape:", goog.shape)
print("NFLX shape:", nflx.shape)
```

Each source file contains 251 observations and 6 original columns. The next cell displays the first few rows of one file so that the structure can be checked.

```python
aapl.head()
```

## Part 4. Add a `ticker` Column




```python
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
goog["ticker"] = "GOOG"
nflx["ticker"] = "NFLX"

aapl.head()
```

## Part 5. Combine the Five DataFrames





```python
faanf = pd.concat(
    [fb, aapl, amzn, nflx, goog],
    ignore_index=True
)

print("Combined DataFrame shape:", faanf.shape)
faanf.head()
```

The combined DataFrame should have **1,255 rows and 7 columns**:

`date`, `open`, `high`, `low`, `close`, `volume`, and `ticker`.

## Part 6. Save the Combined Data as `faanf.csv`



```python
faanf_path = DATA_DIR / "faanf.csv"

faanf.to_csv(faanf_path, index=False)

print("Saved file:", faanf_path)
print("File exists:", faanf_path.exists())
```

## Part 7. Read `faanf.csv` and Convert Data Types

### Assignment Step 2(f)



```python
stock_data = pd.read_csv(faanf_path)

stock_data["date"] = pd.to_datetime(stock_data["date"])
stock_data["volume"] = stock_data["volume"].astype("int64")

stock_data = stock_data.sort_values(
    by=["date", "ticker"]
).reset_index(drop=True)

stock_data.dtypes
```

The expected data types are:

- `date`: `datetime64[ns]`
- `volume`: `int64`

The next cell displays the first rows after sorting by date and ticker.

```python
stock_data.head(10)
```

## Part 8. Find the Seven Rows with the Highest Volume



```python
top_7_volume = stock_data.nlargest(7, "volume")

top_7_volume[
    ["date", "ticker", "volume"]
]
```

## Part 9. Reshape the Data with `melt()`



```python
melted_data = stock_data.melt(
    id_vars=["date", "ticker"],
    value_vars=["open", "high", "low", "close", "volume"],
    var_name="variable",
    value_name="value"
)

print("Original shape:", stock_data.shape)
print("Melted shape:", melted_data.shape)

melted_data.head(15)
```

The original combined data contain **1,255 rows**.

There are five measurement columns being melted, so the long-format DataFrame contains:

**1,255 × 5 = 6,275 rows**

Therefore, the expected melted shape is **6,275 rows × 4 columns**.

## Part 10. Final Verification


```python
assert stock_data.shape == (1255, 7)
assert pd.api.types.is_datetime64_any_dtype(stock_data["date"])
assert pd.api.types.is_integer_dtype(stock_data["volume"])
assert set(stock_data["ticker"].unique()) == {
    "AAPL", "AMZN", "FB", "GOOG", "NFLX"
}
assert melted_data.shape == (6275, 4)
assert faanf_path.exists()

print("All assignment checks passed.")
print("Combined rows:", len(stock_data))
print("Tickers:", sorted(stock_data["ticker"].unique()))
print("Melted rows:", len(melted_data))
print("faanf.csv saved:", faanf_path.exists())
```
