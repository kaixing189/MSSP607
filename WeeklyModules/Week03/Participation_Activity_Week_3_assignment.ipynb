```python
# a. A function's code runs only when you call the function. It does not run when you define the function. When Python sees a def statement, Python makes a function object. Python gives the function object the function name. Python does not run the code inside the function at that moment. Python runs the code when you call the function.

#b. A Python program has one global scope. A Python program can have many local scopes. Python makes a new local scope each time it calls a function. If you call a function five times, Python makes five local scopes. Python creates them one after another. Python also destroys them one after another.

#c. A local scope holds its variables only while the function runs. When the function returns, Python destroys the local scope. The variables in that local scope are gone. You cannot use them outside the function. If you try, Python gives a NameError. One special case exists. A function can return a variable's value. You can store that value in the global scope. Then the value stays. The original local variable still goes away.

#d. You can stop a program from crashing when it gets an error. You use try and except statements. You put code that may cause an error in a try block. You put code that handles the error in an except block. For example, you can try to convert user input to an integer. If the input is not a number, Python gives a ValueError. The except block catches that error. Then the program prints a message.

```

```python
def collatz(number):
    if number % 2 == 0:
        # If number is even
        result = number // 2
        print(result, end=' ')
        return result
    else:
        # If number is odd
        result = 3 * number + 1
        print(result, end=' ')
        return result

print("Enter number:")
try:
    num = int(input())

    print(num, end=' ')

    while num != 1:
        num = collatz(num)
    print()
except ValueError:
    print("You must enter an integer.")
```
