```python
def collatz(number):
    if number % 2 == 0:
        # If number is even
        result = number // 2
        print(result, end=' ')
        return result
    else:
        # If number is odd
        result = 3 * number + 1
        print(result, end=' ')
        return result

print("Enter number:")
try:
    num = int(input())

    print(num, end=' ')

    while num != 1:
        num = collatz(num)
    print()
except ValueError:
    print("You must enter an integer.")
```
