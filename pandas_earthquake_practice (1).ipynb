````python
import pandas as pd


file_path = '/content/earthquake_project/data/parsed.csv'



df = pd.read_csv(file_path)
df.head()
````

## 2(a) 95th percentile of earthquake magnitude in Japan using `magType == 'mb'`


````python
japan_mb = df[(df['parsed_place'] == 'Japan') & (df['magType'] == 'mb')]
japan_95 = japan_mb['mag'].quantile(0.95)
print('95th percentile:', japan_95)

````

**Answer:** The 95th percentile of `mb` earthquake magnitude in Japan is **4.9**.


## 2(b) Percentage of earthquakes in Indonesia coupled with tsunamis


````python
indonesia = df[df['parsed_place'] == 'Indonesia']
indonesia_tsunami_pct = indonesia['tsunami'].mean() * 100

print('Total Indonesia earthquakes:', len(indonesia))
print('Indonesia earthquakes with tsunami:', indonesia['tsunami'].sum())
print(f'Percentage with tsunami: {indonesia_tsunami_pct:.2f}%')

````

**Answer:** 34 of 147 Indonesian earthquakes were coupled with tsunamis, or approximately **23.13%**.


## 2(c) Summary statistics for earthquakes in Nevada


````python
nevada = df[df['parsed_place'] == 'Nevada']
nevada.describe()
````

There are **681** Nevada earthquake observations. For the `mag` column specifically: mean ≈ **0.5001**, standard deviation ≈ **0.6967**, minimum = **-0.5**, median = **0.4**, and maximum = **2.9**.


## 2(d) Add a Ring of Fire indicator column


````python
ring_of_fire_locations = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala',
    'Mexico', 'Japan', 'Philippines', 'Indonesia', 'New Zealand',
    'Canada', 'Fiji', 'Alaska', 'Washington', 'California', 'Russia',
    'Taiwan', 'Tonga', 'Kermadec Islands'
]


df['ring_of_fire'] = (
    df['parsed_place'].isin(ring_of_fire_locations)
    | df['parsed_place'].str.contains('Antarctic', case=False, na=False)
)

df[['parsed_place', 'ring_of_fire']].head()
````

## 2(e) Number of earthquakes inside and outside Ring of Fire locations


````python
ring_counts = df['ring_of_fire'].value_counts()
print(ring_counts)

print('In Ring of Fire:', df['ring_of_fire'].sum())
print('Outside Ring of Fire:', (~df['ring_of_fire']).sum())

````

**Answer:** **7,188** earthquakes occurred in the specified Ring of Fire locations, and **2,144** occurred outside them.


## 2(f) Tsunami count along the Ring of Fire


````python
ring_tsunami_count = df.loc[df['ring_of_fire'], 'tsunami'].sum()
print('Tsunami count along the Ring of Fire:', ring_tsunami_count)

````

**Answer:** The tsunami count along the specified Ring of Fire locations is **45**.
